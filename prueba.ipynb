```python
import pandas as pd
```

```python
leads_contacts = pd.read_csv('Leads_Contacts.csv', sep=';')
```

```python
top_residencias = (
    leads_contacts[leads_contacts["Tipo de registro"] == "Leads"]
    .groupby("Residencias_interes_corregido")["Correo electrónico"]
    .nunique()
    .sort_values(ascending=False)
    .head(3)
    .reset_index()
    .rename(columns={"Residencias_interes_corregido": "Residencia", "Correo electrónico": "Leads"})
)
resultado = top_residencias
```
